# CLEAN

## Date

```python
date = df['Date'].str.replace('-', '/')
date = pd.to_datetime(date, format='%m/%d/%Y')
df['Date'] = date
```

## Conversion Rate

```python
df["Conversion Rate"] = df["Conversion Rate"].str.rstrip('%').astype(float) / 100
```

# TRANSFORM

## Ad total

```python
AD_COLS = ['Ad Opportunities', 'Ad Requests', 'Ad Responses', 'Ad Assets Received']
```

```python
df["Ad Total"] = df[AD_COLS].sum(axis=1)
```
